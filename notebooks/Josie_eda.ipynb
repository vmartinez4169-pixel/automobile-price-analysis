```python
# Loading the raw dataset
df = pd.read_csv("../data/processed/automobile_cleaned.csv")

#Printing out the columns, count of the non-null values, and the data type
df.info()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Engine size vs. price
sns.scatterplot(data=df, x='engine_size', y='price')
plt.title('Engine Size vs Price')
plt.show()
```

```python
# Horsepower vs. Price
sns.scatterplot(data=df, x='horsepower', y='price')
plt.title('Horsepower vs Price')
plt.show()
```

```python
# Average price by body style
sns.barplot(data=df, x='body_style', y='price')
plt.title('Average Price by Body Style')
plt.show()
```

```python
# Number of cars by fuel type
sns.countplot(data=df, x='fuel_type')
plt.title('Cars by Fuel Type')
plt.show()
```

```python
# Price distribution
sns.histplot(data=df, x='price')
plt.title('Distribution of Car Prices')
plt.show()
```
