```python
import pandas as pd
import numpy as np
```

```python
headers = [
    "symboling", "normalized-losses", "make", "fuel-type", "aspiration",
    "num-of-doors", "body-style", "drive-wheels", "engine-location", "wheel-base",
    "length", "width", "height", "curb-weight", "engine-type", "num-of-cylinders",
    "engine-size", "fuel-system", "bore", "stroke", "compression-ratio",
    "horsepower", "peak-rpm", "city-mpg", "highway-mpg", "price"
]

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/autos/imports-85.data"
df = pd.read_csv(url, names=headers, na_values="?")

# 1. Clean target variable 
df = df.dropna(subset=["price"])
df["price"] = df["price"].astype(float)

```

```python
# 2. Replacing missing blank numbers
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
```

```python
# 3. Replace missing blank text 
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# 4. One-Hot Encode categorical variables
df_clean = pd.get_dummies(df, columns=categorical_cols, drop_first=True, dtype=int)
```

```python
# 5. Saving cleaned dataset 
df_clean.to_csv('../data/automobile_clean.csv', index=False)
print("Clean dataset saved to ../data/automobile_clean.csv")
```
